```python
%load_ext autoreload
%autoreload 2
from locallib.picarrodb import *
from locallib.query import *
```

```python
# Get all the tables in the database
f = open('ProdDB.py','w')

# Query to get all tables
f.write('from locallib.picarrodb import * \n')
table_query = Query("""SELECT TABLE_NAME FROM INFORMATION_SCHEMA.TABLES WHERE TABLE_TYPE = 'BASE TABLE' ORDER BY TABLE_NAME""")
df_table = table_query.execute(EU1_Conn)

print(f"Found {len(df_table)} tables to process")

for table_idx, table_row in df_table.iterrows():
    table_name = table_row['TABLE_NAME']
    print(f"Processing table {table_idx + 1}/{len(df_table)}: {table_name}")
    
    # Write table comment and initialization
    f.write(f"#Table {table_idx + 1}/{len(df_table)}: {table_name}\n")
    f.write(f'{table_name} = DBTable(\'{table_name}\')\n')
    
    # Query to get all fields/columns for this specific table
    fields_query = Query(f"""
        SELECT 
            COLUMN_NAME,
            DATA_TYPE,
            IS_NULLABLE,
            COLUMN_DEFAULT,
            CHARACTER_MAXIMUM_LENGTH
        FROM INFORMATION_SCHEMA.COLUMNS 
        WHERE TABLE_NAME = '{table_name}'
        ORDER BY ORDINAL_POSITION
    """)
    
    try:
        df_columns = fields_query.execute(EU1_Conn)
        print(f"  Found {len(df_columns)} columns for {table_name}")
        
        # Process each column with different variable names to avoid conflicts
        for col_idx, column_row in df_columns.iterrows():
            column_name = column_row['COLUMN_NAME']
            column_type = column_row['DATA_TYPE']
            f.write(f"{table_name}.add_column(DBColumn('{column_name}', datatype='{column_type}'))\n")
            
    except Exception as e:
        print(f"  ERROR processing columns for {table_name}: {e}")
        f.write(f"# ERROR: Could not process columns for {table_name}: {e}\n")
    
    f.write('\n')  # Add blank line between tables

f.close()
print("Database schema generation completed!")
```
